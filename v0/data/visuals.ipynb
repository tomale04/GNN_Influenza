```python
import networkx as nx
import matplotlib.pyplot as plt

# --------------------------------------------------
# Original graph
# --------------------------------------------------
example_tree = nx.DiGraph()

nodes = [1, 2, 3, 4, 5, 6, 7, 8, 9]
edges = [(1,2), (2,3), (2,4), (3,5), (3,6), (1,7), (7,8), (7,9)]
edge_colours = ["grey", "grey", "red", "grey", "grey", "grey", "grey", "grey"]

example_tree.add_nodes_from(nodes)
example_tree.add_edges_from(edges)

# --------------------------------------------------
# Graph after contraction
# --------------------------------------------------
contracted_tree = nx.DiGraph()

nodes = [1, 23, 4, 5, 6, 7, 8, 9]
edges = [(1,23), (23,4), (23,5), (23,6), (1,7), (7,8), (7,9)]
contracted_edge_colours = ["grey", "grey", "grey", "grey", "grey", "grey", "grey"]

contracted_tree.add_nodes_from(nodes)
contracted_tree.add_edges_from(edges)

# --------------------------------------------------
# Plotting
# --------------------------------------------------
fig, axes = plt.subplots(1, 2, figsize=(11, 5))


graphs = [
    (example_tree, edge_colours, "Before Contraction"),
    (contracted_tree, contracted_edge_colours, "After Contraction")
]

for ax, (graph, colours, title) in zip(axes, graphs):

    pos = nx.nx_agraph.graphviz_layout(graph, prog="dot")

    nx.draw_networkx_edges(
        graph,
        pos,
        ax=ax,
        edge_color=colours,
        width=2.0,
        arrows=False
    )

    nx.draw_networkx_nodes(
        graph,
        pos,
        ax=ax,
        node_size=950,
        node_color="#DCEAF7",
        edgecolors="#315A78",
        linewidths=1.5
    )

    nx.draw_networkx_labels(
        graph,
        pos,
        ax=ax,
        font_size=11,
        font_weight="semibold",
        font_color="#243447"
    )

    ax.set_title(title, fontsize=13, fontweight="semibold", pad=15)
    ax.axis("off")
    ax.margins(0.15)

plt.tight_layout()

# Save report-quality versions
fig.savefig("figures/edge_contraction.png", dpi=300, bbox_inches="tight")

plt.show()
```
